```python
import pandas as pd
df = pd.read_csv("data/raw/tables/PSI-25/PSI-25_Experimental table_BASS-II.csv",
                 encoding="utf-8-sig")
df.columns = df.columns.str.replace(r"\s+", " ", regex=True).str.strip()
df["material"] = df["Fuel Sample Material"].str.lower().str.replace(r"\s+", " ", regex=True).str.strip()
df["o2"] = pd.to_numeric(df["Calibrated initial O2 % by vol"], errors="coerce")
flow_txt = df["Fan display"].astype(str) + " | " + df["Air display"].astype(str)
df["has_ext"] = flow_txt.str.contains(r"ext|blowoff", case=False)

print(df.shape)
print(df["material"].value_counts().head(15))       # Q1
print(df.groupby("PI")["o2"].describe())            # Q2 + Q5
print(df["has_ext"].value_counts())                 # Q4
df[["Test #", "PI", "material", "o2", "Fan display", "Air display"]].head(30)  # Q3: eyeball it
```
